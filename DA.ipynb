```python
import pandas as pd
df = pd.read_csv("customer_shopping_behavior.csv")
```

```python
df['Review Rating'] = df.groupby('Category')['Review Rating'].transform(lambda x: x.fillna(x.median()))
```

```python
df.isnull().sum()
```

```python
# Convert column names to lowercase
df.columns = df.columns.str.lower()

# Replace multiple underscores with a single underscore
df.columns = df.columns.str.replace(r'_+', '_', regex=True)

# Remove underscores from the beginning and end
df.columns = df.columns.str.strip('_')

# Replace spaces with underscores
df.columns = df.columns.str.replace(' ', '_')

# Rename purchase amount column
df = df.rename(columns={
    'purchase_amount_(usd)': 'purchase_amount'
})
```

```python
df.columns
```

```python
#create a column to new age grou
labels=['Young Adult','Adult','Middle Age','Senior']
df['age_group']=pd.qcut(df['age'],q=4,labels=labels)
```

```python
df[['age','age_group']].head(10)
```

```python
# create column purchase frequency days
frequency_mapping={
    'Fortnightly':14,
    'Weekly':7,
    'Monthly':30,
    'Quaterly':90,
    'Biweekly':14,
    'Annually':365,
    'Every 3 months':90
}
df['purchase_frequency_days'] = df['frequency_of_purchase'].map(frequency_mapping)
```
